```python
from dotenv import load_dotenv
load_dotenv('../.env')
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model='o4-mini')
print(model.invoke('Oi, eu sou o Goku!').content)
```

```python
def add(x: int, y: int) -> int:
    """
    Make the add math operation
    Args:
        x (int) -> The first number to be sum
        y (int) -> The sec number to be sum
    Return:
        result (int) - > The math operation of 'x + y'
    """
    result = x+y
    return result

def subtract(x: int, y: int) -> int:
    """
    Make the subtract math operation
    Args:
        x (int) -> The first number to be subtracted
        y (int) -> The sec number to be subtracted
    Return:
        result (int) - > The math operation of 'x - y'
    """
    result = x-y
    return result
```

```python
from langgraph.prebuilt import create_react_agent
from langgraph.checkpoint.memory import InMemorySaver

checkpointer = InMemorySaver()

math_agent = create_react_agent(
    model=model,
    #model = "openai:o4-mini"
    tools=[add, subtract],
    prompt="You are a math assistant.",
    checkpointer=checkpointer
)
```

```python
# Run the agent
config = {"configurable": {"thread_id": "1"}}

add_response = math_agent.invoke(
    {"messages": [{"role": "user", "content": "quanto é 1+2?"}]},
    config   # type: ignore
)

sub_response = math_agent.invoke(
    {"messages": [{"role": "user", "content": "quanto é 3-1?"}]},
    config # type: ignore
)
```

```python
print(f"Add response: {add_response['messages']}")

print(f"Sub response: {sub_response['messages']}")
```
